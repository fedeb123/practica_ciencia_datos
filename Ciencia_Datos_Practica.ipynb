```python
import numpy as np
import pandas as pd

dataset = pd.read_csv("./bank.csv")

#Ejercicio 2

personas_depositos = (dataset['deposit']).value_counts()
#len da el numero entero
#value_counts() muestra el conteo de valores entre sus diferentes posibilidades de acuerdo a los valores vistos en las columnas

saldo_promedio_por_trabajo = dataset.groupby('job')['balance'].mean().sort_values(ascending = False)
#muy parecido a SQL

personas_con_hipotecas = (dataset['housing'] == 'yes').mean() * 100
# como true vale 1 y false 0, al sacar el promedio se obtiene el porcentaje decimal

saldos_negativos = dataset[dataset['balance'] < 0]['job'].value_counts()
# dataset['balance < 0] devuelve un dataset con true y false segun la condicion, ojo.
# dataset[dataset['balance'] < 0] devuelve un dataset identico pero solo con los saldos negativos
#agregarle ['job'].value_counts() contara cada posibilidad diferente de la columna job sumando y agrupando por posibilida de valor

print(personas_depositos)

print(saldo_promedio_por_trabajo)

print(f"{personas_con_hipotecas}%")

print(saldos_negativos)
```
